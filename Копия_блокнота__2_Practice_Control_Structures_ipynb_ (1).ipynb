- **Cüt və ya Tək Rəqəm Yoxlanışı (If/Else)
Tapşırıq**
-check_even_odd(num) adlı bir funksiya yaradın:
- Parametr kimi bir tam ədəd qəbul etsin.
- If/Else vasitəsi ilə ədədin cüt və ya tək olduğunu yoxlasın.
- Cüt olarsa "Cüt", tək olarsa "Tək" qaytarsın (və ya çap etsin)



```python
def check_even_odd(num):
    if num % 2 == 0:
        return "Cüt"
    else:
        return "Tək"

# Funksiyanı sınaqdan keçirmək
test_number = 37
result = check_even_odd(test_number)
print(f"{test_number} ədədi {result}-dir")
```

**while Dövrü**


(Mövzuya dərsdə toxunulub və tələbənin araşdırma apararaq inkişaf etmə, problem həll etmə qabiliyyətini inkişaf etdirməsi üçün qoyulub.)

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

İzah

i = 1 → başlanğıc dəyər

while i <= 5: → şərt doğru olduğu müddətdə dövr işləyir

print(i) → i ekrana çıxarılır

i += 1 → i hər dövrdə 1 artır

i 6 olanda şərt yalnış olur və dövr dayanır

# Bu hissədə while dövrünü öyrənirəm. while dövrü verilən şərt doğru olduğu müddətdə eyni kodu təkrar-təkrar işlətmək üçün istifadə olunur.

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

# Bu nümunədə i dəyişəni əvvəlcə 1 olaraq təyin olunur. while dövrü i 5-dən kiçik və ya bərabər olduğu müddətdə işləyir. Hər dövrdə i ekrana çıxarılır və i += 1 ilə 1 vahid artırılır. i 6 olduqda şərt yanlış olur və dövr dayanır.

2.**while Dövrü ilə Toplama**
- sum_upto_n(n) adlı funksiya yazın:
- 1-dən n-ə qədər (n daxil olmaqla) bütün ədədləri toplayın (bunun üçün while dövründən istifadə edin).
- Toplamı qaytarsın.
- n mənfi olarsa, 0 qaytarın və ya xəbərdarlıq mesajı çap edin (öz seçiminizdir).

```python
def sum_upto_n(n):
    if n < 0:
        return 0

    total = 0
    i = 1

    while i <= n:
        total += i
        i += 1

    return total
```

```python
# Test
test_values = [5, 10, -3, 0]

for val in test_values:
    print(f"sum_upto_n({val}) = {sum_upto_n(val)}")
```

# Bu funksiyada əvvəl n mənfidirsə, 0 qaytarılır. Əgər n mənfi deyilsə, total dəyişəni yaradılır və 1-dən başlayaraq n-ə qədər olan ədədlər while dövrü ilə total-a əlavə olunur. Sonda toplam nəticə return ilə qaytarılır.

3. **FizzBuzz**
- fizz_buzz(n) adlı bir funksiya yaradın və 1-dən n-ə qədər dövr edin:
- Əgər ədəd 3-ə bölünürsə, "Fizz",
- Əgər ədəd 5-ə bölünürsə, "Buzz",
- Əgər ədəd həm 3-ə, həm də 5-ə bölünürsə, "FizzBuzz",
- Əks halda, ədədin özünü çap edin.

```python
def fizz_buzz(n):
    i = 1

    while i <= n:
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)

        i += 1

fizz_buzz(15)
```

# 3. FizzBuzz Bu tapşırıqda `fizz_buzz(n)` adlı funksiya yaradılır. Funksiya 1-dən n-ə qədər olan ədədləri yoxlayır.
##  Əgər ədəd həm 3-ə, həm də 5-ə bölünürsə, ekrana `FizzBuzz` çıxır.  
## Əgər yalnız 3-ə bölünürsə, `Fizz` çıxır.  
## Əgər yalnız 5-ə bölünürsə, `Buzz` çıxır.  
## Əgər bu şərtlərin heç biri ödənmirsə, ədədin özü çap olunur.

### Bu tapşırıqda `while`, `if`, `elif`, `else` və `%` operatorundan istifadə olunur.

**Saitlərin Sayını Tapmaq**
- count_vowels(text) adlı funksiya yazın:
- Giriş kimi sətir (string) qəbul etsin.
- Saitlərin (a, e, i, o, u və eyni zamanda böyük hərflər) sayını - hesablamaq üçün dövrdən istifadə etsin.


```python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for letter in text:
        if letter in vowels:
            count += 1

    return count

# Test
test_strings = ["Hello", "Education", "Rhythm", ""]
for t in test_strings:
    print(f"'{t}' sözündə saitlərin sayı: {count_vowels(t)}")
```

# vowels dəyişənində sait hərfləri saxladım. count əvvəl 0-dır. for dövrü ilə text içindəki hər hərfə baxdım. Əgər hərf saitlərin içində varsa, count 1 artırılır. Sonda count nəticə kimi qaytarılır.

#5.**Sətiri Tərsinə Çevirmək**
- reverse_string(text) adlı funksiya yaradın:
- Giriş kimi sətir qəbul etsin.
- Verilən sətirin tərsini qaytarsın.
- Dövr və ya text[::-1] kimi qısa üsullardan istifadə edə bilərsiniz.

```python
def reverse_string(text):

    return text[::-1]

# Test
test_words = ["Python", "Colab", "Function", "12345"]
for w in test_words:
    print(f"{w} sətirinin tərsi -> {reverse_string(w)}")
```

# text[::-1] sətiri sondan əvvələ doğru oxuyur. Buna görə verilən söz və ya rəqəm ardıcıllığı tərsinə çevrilir.
